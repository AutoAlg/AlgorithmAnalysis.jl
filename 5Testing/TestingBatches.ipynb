```python
import subprocess
import pyautogui
import time
import os

# Define paths with correct escape handling
julia_script_path = "S:\\Research Material\\CSE\\MU\\Project\\1Code\\AlgorithmAnalysis.jl\\1CPU\\src\\startupCPU.jl"
output_file_path = "S:\\Research Material\\CSE\\MU\\Project\\5Testing\\output.txt"

# Start a new PowerShell window
process = subprocess.Popen('powershell.exe', creationflags=subprocess.CREATE_NEW_CONSOLE)
time.sleep(2)  # Wait for the PowerShell window to open

# Send the command to start the Julia REPL
pyautogui.typewrite('julia')
pyautogui.press('enter')
time.sleep(2)  # Give time for Julia to start

def run_julia_script(run_index):
    # Send the command to run the Julia script with the corrected path
    pyautogui.typewrite(f'include("S:\\\\Research Material\\\\CSE\\\\MU\\\\Project\\\\1Code\\\\AlgorithmAnalysis.jl\\\\1CPU\\\\src\\\\startupCPU.jl")')
    pyautogui.press('enter')
    time.sleep(5)  # Adjust based on how long the script runs

    # Check if the output file exists, wait until it's created
    while not os.path.exists(output_file_path):
        print("Waiting for Julia to create the output file...")
        time.sleep(1)

    # Read output from the file
    with open(output_file_path, 'r') as file:
        output = file.read()

    # Display the output
    print(f"Output from run {run_index}:")
    print(output)

    # Delete the file after reading
    os.remove(output_file_path)

# Run the Julia script multiple times in the same Julia session
for i in range(3):
    run_julia_script(i + 1)

# Send the exit command to close the Julia REPL
pyautogui.typewrite('exit()')
pyautogui.press('enter')
```
